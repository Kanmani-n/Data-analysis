                                    SLICING AN ARRAY IN PYTHON
Slicing an array is the concept of cutting out a part of an array which is fulfilled by Numpy in Python.
Sliced Part is then stored in other variable and can be further modified as required.
Array in python can be sliced using slice() built in function

```python
import numpy as np
```

```python
#creating array
arr_s = np.array([5, 10, 30, 42, 56, 98 ]) 
```

```python
#slicing
s1 = slice(1, 4)
sliced_part = arr_s[s1]
print("sliced array is ",sliced_part)
```
