```python
#2.Find the maximum and minimum elements in a list without using max() or min().
def elements(i):
    s = sorted(i)
    return s[-1], s[0]
```

```python
#5.Merge two sorted lists into a single sorted list
def merge(m1,m2):
    m3=[]
    i=0
    j=0
    while i<len(m1) and j <len(m2):
        if m1[i]<m2[j]:
            m3.append(m1[i])
            i+=1
        else:
            m3.append(m2[j])
            j+=1
    m3.extend(m1[i:])
    m3.extend(m2[j:])
    return m3
            
```

```python
m1=[2,9,6]
m2=[4,8,1]
print(merge(m1,m2))
```

```python
#11. Count the frequency of each character in a string using a dictionary.
def character(str):
    f={}
    for i in str:
        if i in f:
            f[i]+=1
        else:
            f[i]=1
    return f
```

```python
r="kanmani"
s=character(r)
print(s)
```
