Classification is a supervised machine learning technique used to predict a categorical class label. The model learns from labeled data and classifies new data points into one of the predefined categories.
Types of Classification:
Binary Classification: Only two possible classes (e.g., 0 or 1)
Multi-class Classification: More than two classes (e.g., classify animals as Cat, Dog, or Bird)
Multi-label Classification: Each sample can belong to more than one class

Definition:
Accuracy is the ratio of correct predictions (both true positives and true negatives) to the total number of predictions.
